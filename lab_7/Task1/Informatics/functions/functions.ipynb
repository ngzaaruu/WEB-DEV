```python
def min4(a, b, c, d):
    m = a
    if b < m:
        m = b
    if c < m:
        m = c
    if d < m:
        m = d
    return m


a, b, c, d = map(int, input().split())

print(min4(a, b, c, d))
```

```python
n = int(input())
a = list(map(int, input().split()))

for i in range(1, n):
    if a[i] * a[i - 1] > 0:
        print("YES")
        break
else:
    print("NO")
```

```python
def xor(x, y):
    return (x == 1 and y == 0) or (x == 0 and y == 1)


x, y = map(int, input().split())

print(1 if xor(x, y) else 0)
```
